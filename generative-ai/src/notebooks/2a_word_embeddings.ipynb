## Encoding vs Embeddings
**Encoding:** We need to represent our non-numerical, multimodal data as numbers so we can create models out of them.  
**Embeddings** We need a way to store the data we have encoded and have the ability to perform mathematical functions in an optimized way on them. We store encodings as vectors, usually floating-point representations.

![](images/embedding_method_solution_space.png)

```python
# !pip install scikit-learn

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
import numpy as np


corpus = [
    "cat sat on the mat",
    "the dog chased the rabbit",
    "cat ate fish",
    "dogs and cats are friends",
    "a fish swims in the pond",
    "beer loves to sleep on the grass",
    "lion asked another lion",
    "the music is loud"
]

print("Corpus:")
for i, doc in enumerate(corpus, 1):
    print(f"{i}. {doc}")
```

# Count based methods

## Bag of words

```python
count_vec = CountVectorizer()
X_multi = count_vec.fit_transform(corpus)

df_multi = pd.DataFrame(X_multi.toarray(), columns=count_vec.get_feature_names_out())
print("Document-Term Matrix (Multi-hot):")
df_multi
```

## TF-IDF vectorization

```python
# ---------------------------------------------------------------
# 4. TF–IDF document-term matrix
# ---------------------------------------------------------------
tfidf_vec = TfidfVectorizer()
X_tfidf = tfidf_vec.fit_transform(corpus)

df_tfidf = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vec.get_feature_names_out())
print("Document-Term Matrix (TF-IDF):")
df_tfidf.round(3)
```

## Compare document similarity


```python

from sklearn.metrics.pairwise import cosine_similarity


# Compare 2 documents
doc_0 = [corpus[0]]  # cat sat on the mat
doc_2 = [corpus[2]] # cat ate fish
doc_7 = [corpus[7]] # the music is loud

# Vectorize both ways
multi_hot_pair = count_vec.transform(doc_0 + doc_2 + doc_7)
tfidf_pair = tfidf_vec.transform(doc_0 + doc_2 + doc_7)

sim_multi = cosine_similarity(multi_hot_pair)
sim_tfidf = cosine_similarity(tfidf_pair)


pd.DataFrame([[round(sim_multi[0,1],3), round(sim_multi[0,2],3)], [round(sim_tfidf[0,1],3), round(sim_tfidf[0,2], 3)]])

print(f"CountVectorizer(overlapping cat vs overlapping(the)): {round(sim_multi[0,1],3), round(sim_multi[0,2],3)}")

print(pd.DataFrame(sim_multi))

print(f"TF-IDF(overlapping cat vs overlapping(the)): {round(sim_tfidf[0,1],3), round(sim_tfidf[0,2], 3)}")
print(pd.DataFrame(sim_tfidf))
```

### Check pairwise similarity

```python

pairs = [
    ('car', 'minivan'),   # a minivan is a kind of car
    ('car', 'bicycle'),   # still a wheeled vehicle
    ('car', 'airplane'),  # ok, no wheels, but still a vehicle
    ('car', 'cereal'),    # ... and so on
    ('car', 'communism'),
]
for w1, w2 in pairs:
    print(f"{w1}|{w2}|{wv.similarity(w1, w2)}")
```

### Answer analogy questions

```python

# king + woman - man = ?
# man:woman::king:?
print(wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=2))

# wv['Timbers'] - wv['Portland'] + wv['Seattle'] = ?
print(wv.most_similar(positive=['Timbers', 'Seattle'], negative=['Portland'], topn=2))

# wv['Einstein'] - wv['physics'] + wv['classical_music'] = ?
print(wv.most_similar(positive=['Einstein', 'classical_music'], negative=['physics'], topn=2))
```

### Train your own word2vec model: using gensim

```python
import re
al_regex = re.compile(r"[^a-zA-Z]")
class MyCorpus:
    """An iterator that yields sentences (lists of str)."""

    def __iter__(self):
        corpus_path = "data/mahabharat_gutenberg_lemmatized_sents.txt"
        with open(corpus_path) as fp:
            for line in fp.readlines():
                tokens = line.split()
                tokens = [al_regex.sub('', token) for token in tokens]
                yield tokens
```

```python
import time
import gensim.models

sentences = MyCorpus()
s_time = time.time()
# gensim word2vec defaults 
# self, sentences=None, corpus_file=None, vector_size=100, alpha=0.025, window=5, min_count=5,
# max_vocab_size=None, sample=1e-3, seed=1, workers=3, min_alpha=0.0001,
# sg=0, hs=0, negative=5, ns_exponent=0.75, cbow_mean=1, hashfxn=hash, epochs=5, null_word=0,
# trim_rule=None, sorted_vocab=1, batch_words=MAX_WORDS_IN_BATCH, compute_loss=False, callbacks=(),
# comment=None, max_final_vocab=None, shrink_windows=True,
mahabharat_model = gensim.models.Word2Vec(sentences=sentences, vector_size=100, alpha=0.025, window=5, epochs=5)
print(f"Time taken to train word2vec model: {time.time() - s_time} seconds")
```

```python
my_wv = mahabharat_model.wv
for index, word in enumerate(my_wv.index_to_key):
    if index == 10:
        break
    print(f"word #{index}/{len(my_wv.index_to_key)} is {word}")
```

```python
# find similar words 
for word in ["Krishna", "mace", "Bhishma", "Drona", "Pandu"]:
    top_similars = [words for words, _ in my_wv.most_similar(word, topn=10)]
    print(f"{word}: {top_similars}")
```

```python
# wv['Arjuna'] - wv['Krishna'] + wv['Duryodhana'] = ?
print(my_wv.most_similar(positive=['Arjuna', 'Duryodhana'], negative=['Krishna'], topn=2))

# wv['Drona'] - wv['Arjuna'] + wv['Drupada'] = ?
print(my_wv.most_similar(positive=['Drona', 'Arjuna'], negative=['Drupada'], topn=2))

# note: due to small size of corpus, analogies won"t be as good as googlenews model
```
