## Symbolic AI
Also known as Good Old-Fashioned AI (GOFAI), is a branch of artificial intelligence that focuses on representing knowledge through symbols and logical rules. It aims to solve problems by **manipulating these symbols in a way that mimics human reasoning and decision-making processes**. It was the dominant paradigm in AI from the 1950s to the late 1980s, and it reached its peak popularity during the expert systems boom of the 1980s.

![](images/ml_paradigm.png)

## Machine learning 
It is a field of study in artificial intelligence concerned with the development and study of **statistical algorithms that can learn from data and generalise to unseen data**, and thus perform tasks without explicit instructions.

## What is learning?
The central problem in machine learning and deep learning is to **meaningfully
transform data: in other words, to learn useful representations of the input data at
hand—representations that get us closer to the expected output**.

![](images/ml_representations.png)

## Deep learning
"Deep" in Deep learning stands for this **idea of successive layers of representations**. How many layers contribute to a model of the data is called the depth of the model. Other appropriate names for the field could have been layered representations learning or hierarchical representations learning.

## Before Deep Learning – A Brief History of Machine Learning
   * Deep learning isn’t always the best or only approach; many classical algorithms are still widely used.
   * Choosing the right algorithm depends on data availability and problem type.
   * Understanding earlier methods helps **avoid overusing deep learning as a “one-size-fits-all” tool**.



2. **Probabilistic Modeling (1950s–Present)**

   * Based on applying **statistics and Bayes’ theorem** to data analysis.
   * **Naive Bayes:** assumes feature independence; simple yet effective for many tasks.
   * **Logistic Regression (LogReg):** a foundational classification algorithm; still used for baseline models.


3. **Early Neural Networks (1950s–1980s)**

   * Early concepts of neural nets existed since the 1950s but lacked efficient training methods.
   * The breakthrough came in the **1980s** with the rediscovery of **backpropagation**, enabling gradient-based training.
   * **Yann LeCun’s LeNet (1989):** first successful CNN, used by the USPS for ZIP code recognition.



4. **Kernel Methods (1990s)**
   * **Support Vector Machines (SVMs):** developed by Vapnik and Cortes; find optimal separating hyperplanes between classes.
   * Introduced the **kernel trick**, allowing high-dimensional mappings without explicit computation.
   * Strong theoretical grounding and interpretability made SVMs dominant for a decade.
   * Limitations: poor scalability and weak performance on perceptual tasks (like images), requiring **manual feature engineering**.


5. **Decision Trees, Random Forests, and Gradient Boosting (2000s–2010s)**

   * **Decision Trees:** intuitive, interpretable flowchart-like models.
   * **Random Forests:** ensembles of trees, robust and broadly effective (“second-best” for most tasks).
   * **Gradient Boosting Machines (GBMs):** iterative ensembles that improve on random forests by correcting previous model errors.
   * GBMs and deep learning became the top-performing methods on **Kaggle** after 2014.

## Generative AI

Generative artificial intelligence (Generative AI, GenAI,[1] or GAI) is a subfield of artificial intelligence that uses generative models **to produce text, images, videos, audio, software code or other forms of data.**  These models learn the underlying patterns and structures of their training data and use them to produce new data based on the input, which often comes in the form of natural language prompts.


|                         | **Generative AI**                                                              | **Predictive AI**                                     | **Traditional AI**                                       |
| ----------------------- | ------------------------------------------------------------------------------ | ----------------------------------------------------- | -------------------------------------------------------- |
| **Core Purpose**        | Creates **new, original content** (text, images, code, music, etc.)            | **Forecasts future outcomes** using historical data   | **Automates and optimizes** predefined, rule-based tasks |
| **Learning Type**       | Typically uses **deep learning** and **unsupervised/self-supervised learning** | Uses **supervised learning** and statistical models   | Often uses **rule-based or supervised learning**         |
| **Examples of Output**  | Essays, code, artwork, designs, music, product prototypes                      | Sales forecasts, risk scores, maintenance predictions | Quality control decisions, automation scripts            |
| **Data Dependence**     | Can generate output beyond existing data                                       | Relies heavily on **historical data patterns**        | Works with **predefined datasets and rules**             |
| **Creativity Level**    | **High** – can produce innovative, original ideas                              | **Low** – limited to existing patterns                | **None** – follows rules and fixed logic                 |
| **Examples (Industry)** | Microsoft Copilot, ChatGPT, DALL·E, Google Gemini                                             | Demand forecasting tools, risk assessment systems     | Robotic process automation, quality inspection AI        |

## Generative AI model family
| Family         | Key Mechanism                 | Typical Domain        | Example Models           |
| -------------- | ----------------------------- | --------------------- | ------------------------ |
| Autoregressive | Generate data sequentially - each new token conditioned on previous ones.            | Text, audio           | GPT, WaveNet             |
| Autoencoding   | Learn to **reconstruct input from a compressed or corrupted representation**.             | Representation        | BERT, AE                 |
| VAE            | Encode inputs as distributions (mean + variance), sample from them to generate new data. | Image, text           | β-VAE, CVAE              |
| GAN            | Compete two networks — Generator (creates fake data) vs Discriminator (detects fakes).         | Image, video          | StyleGAN, CycleGAN       |
| Diffusion      | Gradually add noise to data and learn to reverse that process.              | Image, audio, text    | Stable Diffusion, Imagen |
| Energy-based   | Model data via an energy function or score function representing probability gradients.       | Theoretical / general | NCSN, EBMs               |
