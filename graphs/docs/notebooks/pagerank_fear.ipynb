# PageRank: Who is the Most Fearsome?

Edges mean "is afraid of".  The target receives a fear-vote.

By raw in-degree, Wife and Cockroach tied at 2 each. PageRank breaks it.

```python
# !pip install matplotlib networkx
```

```python
import networkx as nx
import matplotlib.pyplot as plt
```

### Build the family fear graph

```python
G = nx.DiGraph()
edges = [
    ("Husband",   "Wife"),
    ("Kids",      "Husband"),
    ("Kids",      "Wife"),
    ("Husband",   "Cockroach"),
    ("Wife",      "Cockroach"),
    ("Cockroach", "Kids"),
]
G.add_edges_from(edges)
```

### Raw in-degree (naive fear metric)

```python
in_deg = dict(G.in_degree())

print("=" * 50)
print("APPROACH 1: Raw In-Degree (count incoming fear edges)")
print("=" * 50)
for node in sorted(G.nodes()):
    bar = "█" * in_deg[node]
    print(f"   {node:<12}  {in_deg[node]}   {bar}")

print()
print("   👉 Wife and Cockroach are TIED at 2 each.")
print("      In-degree says they're equally fearsome. Not convincing.")
```

![](../../images/indegree_centrality.png)

```python
alpha = 0.85

pr = nx.pagerank(G, alpha=alpha, max_iter=1000, tol=1e-12)

ranked = sorted(pr.items(), key=lambda kv: -kv[1])
max_score = ranked[0][1]

print("=" * 50)
print(f"APPROACH 2: PageRank (damping \u03b1 = {alpha})")
print("=" * 50)
for rank, (node, score) in enumerate(ranked, 1):
    bar_w = int((score / max_score) * 40 + 2)
    crown = " \U0001f451" if rank == 1 else ""
    print(f"   #{rank}  {node:<12}  {score:.6f}   {'\u2588' * bar_w}{crown}")
```
