```python
import networkx as nx
import matplotlib.pyplot as plt

edges = [((4,6,2), (6,2)), ((2,6,2), (6,2)), ((3,4,1), (4,1)),
         ((2,4,-4), (4,-4)), ((4,2,3), (2,3)), ((1,4,7), (4,7)),
         ((1,2,5), (2,5)), ((1,3,2), (3,2))]

# Parsed as (source, target, weight)
edge_list = [(4,6,2), (2,6,2), (3,4,1), (2,4,-4),
             (4,2,3), (1,4,7), (1,2,5), (1,3,2)]

G = nx.DiGraph()

for u, v, w in edge_list:
    G.add_edge(u, v, weight=w)

pos = nx.spring_layout(G)

nodes = nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=700, linewidths=2)
edges_drawn = nx.draw_networkx_edges(G, pos, arrowstyle='->', arrowsize=15, connectionstyle='arc3,rad=0.1')

labels = {(u, v): d['weight'] for u, v, d in G.edges(data=True)}
edge_labels = nx.draw_networkx_edge_labels(G, pos, labels, font_size=10)
node_labels = nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold')

plt.title("Directed Graph")
plt.axis('off')
# plt.tight_layout()
plt.show()
```
